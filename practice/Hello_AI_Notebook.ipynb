## Part 4 — List and Dict Comprehensions

Comprehensions are a compact way to build lists/dicts from a loop. Compare these two equivalent approaches:

```python
# Long way: build a list of squares with a loop
squares_loop = []
for n in range(1, 11):
    squares_loop.append(n ** 2)

# Short way: the same thing as a list comprehension
squares_comprehension = [n ** 2 for n in range(1, 11)]

print(squares_loop)
print(squares_comprehension)
print(squares_loop == squares_comprehension)
```

## Part 5 — Mini Exercise: A Simple Reflex Agent

Let's connect this back to this week's lecture. A **simple reflex agent** acts only on the *current percept* — it has no memory of the past.

Below is a tiny simple reflex agent for a thermostat: it looks at the current temperature and decides whether to heat, cool, or do nothing.

```python
def thermostat_agent(current_temp, target_temp=70, tolerance=2):
    """A simple reflex agent: acts only on the current percept (temperature)."""
    if current_temp < target_temp - tolerance:
        return "heat"
    elif current_temp > target_temp + tolerance:
        return "cool"
    else:
        return "do nothing"

# Try it on a few percepts (temperature readings)
for temp in [65, 70, 75, 80, 60]:
    action = thermostat_agent(temp)
    print(f"Temperature: {temp}°F -> Action: {action}")
```
