## Exercise 1 — Classify the Task Type (5%)

Write a function that returns the correct learning type given two simple facts about a problem: whether it has labels, and whether it has a reward signal.

```python
def classify_task(has_labels, has_reward_signal):
    """Return 'supervised', 'unsupervised', or 'reinforcement'."""
    # TODO: implement this
    match (has_labels, has_reward_signal):
        case (True, False):
            return "supervised"
        case (False, True):
            return "reinforcement"
        case (False, False):
            return "unsupervised"
```

**Self-check:**

```python
assert classify_task(has_labels=True, has_reward_signal=False) == "supervised"
assert classify_task(has_labels=False, has_reward_signal=False) == "unsupervised"
assert classify_task(has_labels=False, has_reward_signal=True) == "reinforcement"
print("All tests passed!")
```

## Exercise 2 — Compute Precision and Recall (6%)

Recall the spam-filter example from Lecture 4: out of 100 emails (5 actually spam), the model caught 4 true spam emails, missed 1, and raised 3 false alarms. Write functions that compute precision and recall from raw counts.

```python
def precision(true_positives, false_positives):
    # TODO: return true_positives / (true_positives + false_positives)
    return true_positives / (true_positives + false_positives)

def recall(true_positives, false_negatives):
    # TODO: return true_positives / (true_positives + false_negatives)
    return true_positives / (true_positives + false_negatives)
    
```

**Self-check (matches the Lecture 4 spam-filter example: TP=4, FP=3, FN=1, TN=92):**

```python
tp, fp, fn = 4, 3, 1
assert round(precision(tp, fp), 2) == 0.57
assert round(recall(tp, fn), 2) == 0.80
print("All tests passed!")
```

## Exercise 3 — Diagnose Overfitting vs. Underfitting (6%)

Write a function that takes a training accuracy and a test accuracy and returns a one-word diagnosis. Use the same three cases from the Lecture 4 “Diagnose the Model” practice slide to check yourself.

```python
def diagnose(train_acc, test_acc, gap_threshold=0.15):
    """Return 'overfitting', 'underfitting', or 'good fit'.
    Hint: a big gap between train and test accuracy is overfitting.
    Both scores being low (below ~70%) with a small gap is underfitting.
    Both scores being reasonably high and close together is a good fit.
    """
    # TODO: implement this
    a = train_acc
    b = test_acc
    c = gap_threshold
    if a - b > c:
        return "overfitting"
    elif a and b < 0.70:
        return "underfitting"
    elif a - b < c:
        return "good fit"
```

**Self-check (uses the deck's exact Case A / B / C numbers):**

```python
assert diagnose(0.99, 0.61) == "overfitting"   # Case A
assert diagnose(0.64, 0.62) == "underfitting"  # Case B
assert diagnose(0.91, 0.89) == "good fit"      # Case C
print("All tests passed!")
```

## Exercise 4 — Flag Underrepresented Groups (6%)

Extend the representation-imbalance idea from the Exploration Notebook's hiring dataset (school types A, B, and C). Write a function that flags any group making up less than a given fraction of the total.

```python
def underrepresented_groups(counts, min_fraction=0.2):
    """counts is a dict like {'A': 9, 'B': 2, 'C': 1}.
    Return a list of group names whose share of the total is below min_fraction.
    """
    # Get the total sum as a variable
    sumCount = sum(counts.values())
    # Get the cutoff number
    cutoff = sumCount * min_fraction

    # TODO Return a list of type counts whose values are less than min_fraction (20%) of the sum of the total
    result = []
    for key, value in counts.items():
        if value < cutoff:
            result.append(key)

    return result
    
    
```

**Self-check (uses the notebook's exact school_type counts: A=9, B=2, C=1):**

```python
school_type_counts = {"A": 9, "B": 2, "C": 1}
assert sorted(underrepresented_groups(school_type_counts)) == ["B", "C"]
print("All tests passed!")
```

## Exercise 6 — Train / Validation / Test Split Sizes (5%)

Write a function that takes a total dataset size and returns how many examples go into train, validation, and test, using the 70% / 15% / 15% split from Lecture 4.

```python
def split_sizes(total, train_frac=0.70, val_frac=0.15, test_frac=0.15):
    """Return a (train, val, test) tuple of example counts."""
    # TODO takes a total dataset size and returns how many examples go into train, validation, and test, using the 70% / 15% / 15% split
    a = total * train_frac
    b = total * val_frac
    c = total * test_frac
    return (a, b, c)
```

**Self-check (matches the 70% / 15% / 15% split shown in Lecture 4):**

```python
assert split_sizes(100) == (70, 15, 15)
assert split_sizes(200) == (140, 30, 30)
print("All tests passed!")
```

## Exercise 7 — Mean Squared Error Over Several Guesses (6%)

The Exploration Notebook's gradient descent section scores a single guess at a time. Extend that idea: write a function that takes a true value and a list of guesses, and returns the average squared error across all of them.

```python
def mean_squared_error(true_value, guesses):
    """Return the average squared error across a list of guesses."""
    # TODO: implement this
    total = 0
    for guess in guesses:
        difference = (true_value - guess)
        difference = difference ** 2
        total = total + difference
    result = total / len(guesses)
    return result
```

**Self-check (uses the notebook's exact guesses: 70, 85, 95, 100, 110):**

```python
assert mean_squared_error(100, [70, 85, 95, 100, 110]) == 250
print("All tests passed!")
```
